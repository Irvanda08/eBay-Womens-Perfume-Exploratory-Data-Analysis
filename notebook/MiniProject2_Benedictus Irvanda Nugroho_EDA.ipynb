```python
def categorize_type_refined(text):
    text = str(text).lower()
    
    if 'extrait' in text or 'elixir' in text:
        return 'Extrait de Parfum'
    
    # Kelompok Eau de Parfum (EDP) & Parfum
    edp_keywords = ['eau de parfum', 'edp', 'parfum', 'perfume', 'parfume', 'elixir', 'fragrance', 'spray']
    if any(kw in text for kw in edp_keywords):
        return 'Eau de Parfum'
    
    # Kelompok Eau de Toilette (EDT)
    edt_keywords = ['eau de toilette', 'edt', 'toilette']
    if any(kw in text for kw in edt_keywords):
        return 'Eau de Toilette'
    
    # Kelompok Cologne
    cologne_keywords = ['cologne', 'edc']
    if any(kw in text for kw in cologne_keywords):
        return 'Cologne'
    
    # Kelompok Body Mist
    mist_keywords = ['mist', 'body spray', 'fragrance mist', 'shimmer']
    if any(kw in text for kw in mist_keywords):
        return 'Body Mist'
    
    # Kelompok Gift Set
    set_keywords = ['set', 'pc', 'gift', 'asst']
    if any(kw in text for kw in set_keywords):
        return 'Gift Set'

    body_care_keywords = ['lotion', 'cream', 'moisturizer', 'deodorant', 'powder', 'bath oil', 'skin_moisturizer']
    if any(kw in text for kw in body_care_keywords):
        return 'Body Care'
    
    # Kelompok Unknown / Data Sampah
    unknown_keywords = ['does not apply', '/', '1', 'nan', 'beauty']
    if any(kw in text for kw in unknown_keywords) or text.isdigit():
        return 'Unknown'
    
    return 'Others'

# Terapkan ulang
df_clean['type'] = df_clean['type'].apply(categorize_type_refined)
```

#### Pengujian untuk Normalisasi Data

```python
def try_transformations(df, column):
    
    # (tambah 1 untuk menghindari log(0) atau boxcox(0))
    data_asli = df[column] + 1
    
    # METODE 1: SQUARE ROOT
    data_sqrt = np.sqrt(data_asli)
    
    # METODE 2: LOG TRANSFORMATION
    data_log = np.log1p(data_asli)
    
    # METODE 3: BOX-COX
    data_boxcox, lam = stats.boxcox(data_asli)
    
    # List untuk looping visualisasi
    methods = [
        ('Original', data_asli),
        ('Square Root', data_sqrt),
        ('Log', data_log),
        ('Box-Cox', data_boxcox)
    ]
    
    # Visualisasi Perbandingan
    plt.figure(figsize=(20, 5))
    for i, (name, data) in enumerate(methods):
        plt.subplot(1, 4, i+1)
        sns.histplot(data, kde=True, color='teal')
        
        # Hitung Shapiro-Wilk
        stat, p = stats.shapiro(data)
        plt.title(f"{name}\nShapiro p-val: {p:.10f}")
        
        print(f"Metode {name:12} | P-Value: {p:.10f}")
    
    plt.tight_layout()
    plt.show()

kolom_numerik = ['price', 'sold', 'available']

for col in kolom_numerik:
    try_transformations(df_clean_6, col)
```

```python
import re

def extract_size_ml(title):
    title = str(title)
    
    # Tangkap pola "2.5 fl oz" atau "2.5fl oz" (ada kata 'fl' di antara angka dan oz)
    fl_oz = re.search(r'(\d*\.?\d+)\s?fl\.?\s?oz', title, re.IGNORECASE)
    if fl_oz:
        return round(float(fl_oz.group(1)) * 29.57)
    
    # Tangkap pola biasa "50ml" atau "50 ml"
    ml = re.search(r'(\d+)\s?ml', title, re.IGNORECASE)
    if ml:
        return float(ml.group(1))
    
    # Tangkap pola "1.7oz" atau "1.7 oz" (tanpa fl)
    oz = re.search(r'(\d*\.?\d+)\s?oz', title, re.IGNORECASE)
    if oz:
        return round(float(oz.group(1)) * 29.57)
    
    return None  # Tidak ditemukan ukuran

df_manip['size_ml'] = df_manip['title'].apply(extract_size_ml)

df_manip.head(10)
```

```python
import category_encoders as ce
from sklearn.preprocessing import OrdinalEncoder

# One-Hot Encoding : TYPE

ohe_type = pd.get_dummies(
    df_encoded['type'],
    prefix='oh_type',
    dtype=int
)

# One-Hot Encoding : COUNTRY

ohe_country = pd.get_dummies(
    df_encoded['country'],
    prefix='oh_country',
    dtype=int
)

# Gabungkan hasil OHE ke dataframe
df_encoded = pd.concat(
    [df_encoded, ohe_type, ohe_country],
    axis=1
)

# Binary Encoding : BRAND

be = ce.BinaryEncoder(cols=['brand'])

brand_binary = be.fit_transform(df_encoded[['brand']])

# Rename kolom binary encoding
brand_binary.columns = [
    f'be_brand_{i}'
    for i in range(brand_binary.shape[1])
]

# Gabungkan hasil Binary Encoding
df_encoded = pd.concat(
    [df_encoded, brand_binary],
    axis=1
)

# Ordinal Encoding : PRICE CATEGORY

oe = OrdinalEncoder(
    categories=[['Budget', 'Midrange', 'Prestige', 'Luxury']]
)

df_encoded['oe_price_category'] = oe.fit_transform(
    df_encoded[['price_category']]
)

# Menampilkan hasil
df_encoded.head()
```

```python
# Menentukan ambang batas (boundary) menggunakan pendekatan persentil secara dinamis
# Mengambil top 25% produk dengan penjualan tertinggi sebagai indikator "laris"
batas_sold_banyak = df_eda['sold'].quantile(0.75)
# Mengambil bottom 25% sisa stok terkecil sebagai indikator "stok kritis"
batas_stok_menipis = df_eda['available'].quantile(0.25)

# Memfilter data menggunakan logika kondisi ganda (AND)
# Produk harus memenuhi kedua syarat: penjualan di atas batas DAN stok di bawah/sama dengan batas
restock_urgently = df_eda[
    (df_eda['sold'] > batas_sold_banyak) & 
    (df_eda['available'] <= batas_stok_menipis)
]

# Menampilkan informasi ringkas hasil kalkulasi boundary dan jumlah temuan data
print(f"Batas Sold Banyak: > {batas_sold_banyak:.0f} botol")
print(f"Batas Stok Menipis: <= {batas_stok_menipis:.0f} botol")
print(f"Jumlah produk yang harus segera di-restock: {len(restock_urgently)} produk\n")

# Menampilkan 5 sampel produk teratas yang masuk dalam daftar restock mendesak
print("Sampel Produk yang Mengalami Krisis Stok:")
display(restock_urgently[['brand', 'title', 'sold', 'available', 'price', 'revenue']].head())
```

**Insight**

- Terdapat 24 produk yang sedang laris, namun saat ini kondisinya sangat kritis karena stok yang tersisa kurang dari atau sama dengan 5 botol. Jika tidak segera restock, toko akan kehilangan pembeli.
- Hal ini menunjukkan adanya manajemen inventori yang kurang reaktif terhadap produk best-seller. Produk ini laju penjualan nya sangat cepat, namun kuantitas pengadaan stok nya lamban

```python
# Uji hipotesa

# Pengaruh Tipe Parfum terhadap Jumlah Penjualan Produk

# Setiap tipe parfum memiliki karakteristik aroma, konsentrasi, dan ketahanan yang berbeda sehingga dapat mempengaruhi preferensi konsumen dan performa penjualan produk di marketplace.

from scipy.stats import kruskal

# Kelompok sold berdasarkan kategori harga
budget = df_eda[df_eda['price_category'] == 'Budget']['sold']
midrange = df_eda[df_eda['price_category'] == 'Midrange']['sold']
prestige = df_eda[df_eda['price_category'] == 'Prestige']['sold']
luxury = df_eda[df_eda['price_category'] == 'Luxury']['sold']

# Uji Kruskal-Wallis
stat, pvalue = kruskal(
    budget,
    midrange,
    prestige,
    luxury
)

print('Statistic:', stat)
print('P-Value:', pvalue)

print("Kesimpulan:")
if pvalue > 0.05:
    print("Hasil       : Gagal Tolak H0 (P-Value > 0.05)")
    print("Artinya     : Distribusi penjualan ketiga tipe parfum cenderung SAMA (tidak ada perbedaan nyata).")
else:
    print("Hasil       : Tolak H0 (P-Value <= 0.05)")
    print("Artinya     : Artinya terdapat perbedaan signifikan jumlah penjualan antar kategori harga parfum.")
```
